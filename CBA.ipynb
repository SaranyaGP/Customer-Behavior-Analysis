```python
#common actions leading to successful conversions.


sql = """Select
    Action, 
    COUNT(DISTINCT CustomerID) AS total_customers, 
    COUNT(DISTINCT CASE WHEN JourneyID = 'Yes' THEN CustomerID END) AS converted_customers,
    ROUND(
        (COUNT(DISTINCT CASE WHEN JourneyID = 'Yes' THEN CustomerID END) / 
        COUNT(DISTINCT CustomerID)) * 100, 
        2
    ) AS conversion_rate
FROM customer_journey
GROUP BY Action
ORDER BY conversion_rate DESC;"""
```

```python
curr = db_connection.cursor()
```

```python
curr.execute(sql)
```

```python
Conversion_rate = curr.fetchall()
```

```python
curr.column_names
```

```python
Conversion_rate = pd.DataFrame(Conversion_rate,columns = curr.column_names)  
```

```python
Conversion_rate
```
